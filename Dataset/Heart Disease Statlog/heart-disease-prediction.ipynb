```python
# Read data
df = pd.read_csv('/kaggle/input/heart-disease-statlog/Heart_disease_statlog.csv')
df.head(3)
```

```python
df.info()
```

```python
# Define features and target
y = df['target']
X = df.drop('target', axis=1)
```

```python
# Split train and test data
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.3, random_state=12)
```

```python
# Scale data
scaler = StandardScaler()

X_train_scaled = scaler.fit_transform(X_train)
X_test_scaled = scaler.transform(X_test)
```

```python
# Define model
xg = xgb.XGBClassifier()

# Fit model
model = xg.fit(X_train_scaled, y_train)

# Predict outputs
predictions = model.predict(X_test_scaled)
predictions
```

```python
# Define RF model
rf = RandomForestClassifier()

# Fit model
model_rf = rf.fit(X_train_scaled, y_train)

# Predict outputs
predictions_rf = model_rf.predict(X_test_scaled)
predictions_rf
```

```python
# Display model score
model.score(X_test_scaled, y_test)
```

```python
model_rf.score(X_test_scaled, y_test)
```

```python
# Display cross validation scores
cv_scores = cross_val_score(xg, X_train_scaled, y_train, cv=5)  # 5-fold cross validation

print(f"Cross-Validation Scores: {cv_scores}")
print(f"Mean CV Accuracy: {cv_scores.mean():.4f}")
```

```python
cv_scores_rf = cross_val_score(rf, X_train_scaled, y_train, cv=5)  # 5-fold cross validation

print(f"Cross-Validation Scores: {cv_scores_rf}")
print(f"Mean CV Accuracy: {cv_scores_rf.mean():.4f}")
```

```python
# Convert predictions to Dataframe
pred_df = pd.DataFrame({
    'actual': y_test.values,
    'XGB_Prediction': predictions,
    'RF_Prediction': predictions_rf
})

pred_df.to_csv('predictions.csv', index=False)

print("Predictions saved to CSV file successfully.")
```
