## 📊 範例一：KMeans 分群 – Iris 資料集

```
from sklearn.datasets import load_iris
from sklearn.cluster import KMeans
from sklearn.metrics import silhouette_score
import matplotlib.pyplot as plt
import pandas as pd
import seaborn as sns

# 載入資料
iris = load_iris()
X = iris.data
y = iris.target

# 建立 KMeans 模型
kmeans = KMeans(n_clusters=3, random_state=42)
clusters = kmeans.fit_predict(X)

# Silhouette 分數
score = silhouette_score(X, clusters)
print(f"Silhouette Score: {score:.2f}")

# 可視化
df = pd.DataFrame(X, columns=iris.feature_names)
df['Cluster'] = clusters
sns.pairplot(df, hue='Cluster', palette='tab10')
plt.show()
```

## 🌙 範例二：DBSCAN 分群 – 月牙形資料

```
from sklearn.datasets import make_moons
from sklearn.cluster import DBSCAN

# 產生月牙形狀資料
X, y = make_moons(n_samples=300, noise=0.05, random_state=42)

# DBSCAN 聚類
dbscan = DBSCAN(eps=0.2, min_samples=5)
clusters = dbscan.fit_predict(X)

# 可視化
plt.figure(figsize=(8, 4))
plt.scatter(X[:, 0], X[:, 1], c=clusters, cmap='tab10', s=50)
plt.title('DBSCAN 分群結果')
plt.xlabel('X1')
plt.ylabel('X2')
plt.show()
```

## 🖼️ 範例三：KMeans 圖像壓縮

```
from sklearn.cluster import KMeans
from skimage import io
import numpy as np

# 載入圖像
image = io.imread('https://upload.wikimedia.org/wikipedia/commons/thumb/2/24/Lenna.png/256px-Lenna.png')
rows, cols, channels = image.shape
X = image.reshape(-1, 3)

# KMeans 色彩分群
kmeans = KMeans(n_clusters=16, random_state=0).fit(X)
compressed = kmeans.cluster_centers_[kmeans.labels_].reshape(rows, cols, 3).astype('uint8')

# 原圖與壓縮圖比較
fig, ax = plt.subplots(1, 2, figsize=(10, 5))
ax[0].imshow(image)
ax[0].set_title("原圖")
ax[0].axis('off')
ax[1].imshow(compressed)
ax[1].set_title("壓縮後 (16色)")
ax[1].axis('off')
plt.show()
```
